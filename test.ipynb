```python
def parse_ranges(range_string):
    """Parse the range string and return a list of (start, end, duration) tuples"""
    ranges = []
    items = range_string.split(', ')
    
    for item in items:
        item = item.strip()
        if '-' in item:
            # Range like "446-451"
            start, end = map(int, item.split('-'))
            duration = end - start + 1
            ranges.append((start, end, duration))
        else:
            # Single number like "419"
            num = int(item)
            ranges.append((num, num, 1))
    
    return ranges

def create_sorted_table(range_string):
    """Create a sorted table of ranges by duration"""
    ranges = parse_ranges(range_string)
    
    # Sort by duration (descending), then by start frame
    ranges.sort(key=lambda x: (-x[2], x[0]))
    
    print(f"{'Start':<8} {'End':<8} {'Duration':<10} {'Range'}")
    print("-" * 40)
    
    for start, end, duration in ranges:
        if start == end:
            range_str = str(start)
        else:
            range_str = f"{start}-{end}"
        print(f"{start:<8} {end:<8} {duration:<10} {range_str}")

# Your data
data = "419, 446-451, 612, 685-687, 751-759, 820-829, 901-904, 1085-1103, 1225-1227, 1327-1342, 1371-1375, 1445-1446, 1604-1622, 1678-1679, 1812-1814, 2265, 2333-2346, 2553-2562, 2677, 2886-2892, 2952-2958, 2968-2990, 3010-3012, 3125-3127, 3141, 3352-3354, 3677-3681, 3804-3810, 3948-3949, 3973-3974, 4705-4706, 4721-4728, 4889-4901, 4926-4971, 5107-5111, 5206-5212, 5248-5256, 5312-5338, 5364-5365, 5446-5451, 5472-5482, 5503-5520, 5832-5838, 5997-5999, 6022-6028, 6047, 6116-6126, 6283-6308, 6326-6343, 6550, 6558-6568, 6660-6661, 6683-6686, 6798-6802, 7094-7098, 7147-7150, 7275, 7341-7348, 7442-7447, 7477-7525, 7586-7589, 7651-7656, 7910-7915, 8087, 8377-8387, 8453, 8636-8637, 8672, 8694, 8754-8763, 8787-8791, 8858-8867, 8885-8890, 9033, 9053-9056, 9142-9145, 9335-9359, 9389, 9406-9409, 9521"

create_sorted_table(data)
```
