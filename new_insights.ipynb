# Statistical Analysis of Behavioral Data
This notebook uses tools from `statistics.py` to analyze and generate insights about the behavioral data.

```python
import numpy as np
import pandas as pd
import os, re
import matplotlib.pyplot as plt
from src.analysis.statistics import StatisticalAnalysis
from src.analysis.data_model import AnnotationData
from src.analysis.visualization import VisualizationManager
```

```python
# Define paths to data directories
data_path = '/Users/hind/Documents/UCSB/Neuroscience/headcasting project/Filtered Supervised vs Peristalsis/'
peristalsis_path = data_path + 'Persistalsis/'
quantile_50_path = data_path + 'quantile_50_preds/'
```

```python
def extract_number(s):
    numbers = re.findall(r'\d+', s)  # Find all numbers in the string
    return int(numbers[-1])

def load_csv_files(path, trans=False, pc = False):
    csv_files = [os.path.join(root, file) 
             for root, _, files in os.walk(path) 
             for file in files if file.endswith('.csv')]
    if trans:
        file_list = [f for f in csv_files if 'trans' in f and 'mismatch' not in f]
    elif pc:
        file_list = [f for f in csv_files if 'pc' in f and 'mismatch' not in f]
    else:
        file_list = [f for f in csv_files if 'trans' not in f and 'pc' not in f and 'mismatch' not in f]

    file_list.sort(key=extract_number)

    return file_list
```

```python
peristalsis_files = load_csv_files(peristalsis_path)
quantile_50_files = load_csv_files(quantile_50_path)
```

```python
# Initialize AnnotationData objects for peristalsis and quantile_50 datasets
peristalsis_data = AnnotationData()
quantile_50_data = AnnotationData()

# Load annotations from CSV files
peristalsis_data.load_from_file(peristalsis_files[0], column=2)
quantile_50_data.load_from_file(quantile_50_files[0], column=2)

# Display metadata
print("Peristalsis Metadata:", peristalsis_data.metadata)
print("Quantile 50 Metadata:", quantile_50_data.metadata)
```
